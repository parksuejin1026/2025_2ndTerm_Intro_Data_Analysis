```python
# 행정 구역의 전체 인구인 row[2]가 0이면 오류 발생
import csv
import numpy as np
import matplotlib.pyplot as plt

f = open('age.csv', encoding='cp949')
data = csv.reader(f)
header = next(data)

data = list(data)  #추가

min_val = 1
result_name = ''
result = 0

name = input('인구 구조가 알고 싶은 지역의 이름(읍면동 단위)을 입력해주세요 : ')
for row in data :
    for i in range(102):
        row[i+2] = int(row[i+2].replace(',', ''))
    if name in row[0] :
        home = np.array(row[3:], dtype = int) / row[2]

for row in data :
    if name not in row[0] :
        away = np.array(row[3:], dtype = int) / row[2]
        s = np.sum((home - away) ** 2)
        if s < min_val:
            min_val = s
            result_name = row[0]
            result = away

f.close()

print(result_name)
plt.plot(home)
plt.plot(result)
plt.show()
```

```python
import pandas as pd
import numpy as np

index = pd.date_range('1/1/2024', periods=8)
df = pd.DataFrame(np.random.rand(8, 3), index=index, columns= ['A', 'B', 'C'])
df
```

```python
df['D'] = df['A'] / df['B']  # A열의 값을 B열의 값으로 나눈 값을 D열에 저장
df
```

```python
import numpy as np

df['E'] = np.sum(df, axis=1)  #행 우선 계산 값을 E 열에 저장
df
```

```python
df['F'] = np.ones(8)
df.head(3)
```

```python
df1 = df.drop(['E', 'F'], axis='columns')
df1.head()
```

```python
del df1['D']
df1.head()
```

```python
df1.columns = ['a', 'b', 'c']
df1
```

```python
df1.rename(columns={'a': '2020', 'b': '2021'}, inplace=True)
# df1.rename(columns={'a': '2020', 'b': '2023'})
df1.tail()
```
